```python
import csv
```

```python
csv_data = []
input_filename = 'stops.csv'

# Read the CSV file and store its contents in the variable
with open(input_filename, 'r') as file:
    csv_reader = csv.reader(file)
    for row in csv_reader:
        csv_data.append(row)
```

```python
output_data = [row[0].split(',') for row in csv_data]
```

```python
output_csv_path = 'stops_modifed.csv'
with open(output_csv_path, 'w', newline='') as file:
    csv_writer = csv.writer(file)
    
    # Write the header
    csv_writer.writerow(output_data[0])
    
    # Write the data
    csv_writer.writerows(output_data[1:])
```
